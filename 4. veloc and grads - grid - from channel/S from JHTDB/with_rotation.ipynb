```python
def rotmat(theta, phi):
    m = np.zeros((3, 3))
    m[0, 0] = np.sin(theta) * np.cos(phi)
    m[0, 1] = np.sin(theta) * np.sin(phi)
    m[0, 2] = np.cos(theta)
    m[1, 0] = np.cos(theta) * np.cos(phi)
    m[1, 1] = np.cos(theta) * np.sin(phi)
    m[1, 2] = -np.sin(theta)
    m[2, 0] = -np.sin(phi)
    m[2, 1] = np.cos(phi)
    m[2, 2] = 0.
    return m
```

```python
x = np.array([1, 0, 0], dtype=np.float64)
```

```python
np.matmul(x, rotmat(np.pi/2, np.pi/2))
```
